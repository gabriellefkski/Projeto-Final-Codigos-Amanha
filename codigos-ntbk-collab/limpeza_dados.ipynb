```python
import pandas as pd
import matplotlib as plt
import numpy as np
```

```python
df = pd.read_csv("bnf-2018.csv", encoding='latin-1', sep=";") ##MUDAR NOME CSV OU MUDAR O NOME DO SEU CSV
df.columns = df.columns.str.strip()
df.columns= ["data-concessão", "tipo-beneficio", "cid", "cid-descricao", "tipo-concessão", "dt-nasc-benef", "sexo", "zona-benef", "municipio-benef", "vinculos-dep-benef", "filiacao", "UF", "qt-salarios-minimos-rmi"]
df = df.drop(columns=["vinculos-dep-benef"])
```

```python
df.head()
```

```python
print(df.shape) #qt linhas e colunas
print(df.duplicated().sum()) #linhas duplicadas
print(df.isna().sum().sort_values(ascending=False)) #verificar duplicadas
```

```python
df = df.drop_duplicates()

cid_vazio_dic = ["0", "00", "ZERADOS", "000"]
for col in ["cid", "cid-descricao"]:
  df[col] = df[col].str.strip().str.upper()
  df[col] = df[col].replace(r"^\s*$", np.nan, regex=True)
  df[col] = df[col].replace(cid_vazio_dic, np.nan)


df.head()
print(df.isna().sum())
```

```python
df = df.dropna(subset=["cid", "cid-descricao"])
df.shape
df.head()
```
